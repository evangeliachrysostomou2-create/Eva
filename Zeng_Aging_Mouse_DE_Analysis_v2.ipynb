```python
import scanpy as sc

raw_path = cache.get_file_path(
    directory="Zeng-Aging-Mouse-10Xv3",
    file_name="Zeng-Aging-Mouse-10Xv3/raw"
)

adata = sc.read_h5ad(raw_path, backed="r")

print(adata.shape)
```

```python
import numpy as np

donor_info = cell_metadata[["donor_age", "donor_label"]].drop_duplicates()
donor_info = donor_info.sort_values(["donor_age", "donor_label"]).reset_index(drop=True)

print(donor_info.shape)
```

```python
donor_counts = np.zeros((len(donor_info), adata.n_vars), dtype=np.float64)

donor_to_idx = dict(zip(donor_info["donor_label"], donor_info.index))

for X_chunk, start, end in adata.chunked_X(10000):
    labels = cell_metadata.iloc[start:end]["donor_label"].values

    for donor in np.unique(labels):
        idx = donor_to_idx[donor]
        mask = labels == donor
        donor_counts[idx] += np.asarray(X_chunk[mask].sum(axis=0)).ravel()

print(donor_counts.shape)
```

```python
library_size = donor_counts.sum(axis=1)

log_cpm = np.log1p(
    donor_counts / library_size[:, None] * 1_000_000
)

print(log_cpm.shape)
```

```python
from scipy.stats import ttest_ind

age1 = "P53"
age2 = "P54"

idx1 = donor_info["donor_age"] == age1
idx2 = donor_info["donor_age"] == age2

t_stat, p_values = ttest_ind(
    log_cpm[idx1],
    log_cpm[idx2],
    axis=0,
    equal_var=False
)

print("P53 donors:", idx1.sum())
print("P54 donors:", idx2.sum())
print("Genes tested:", len(p_values))
```

```python
from statsmodels.stats.multitest import multipletests

fdr = multipletests(p_values, method="fdr_bh")[1]

print("Significant genes (FDR < 0.05):", (fdr < 0.05).sum())
```

```python
valid_ages = [
    age for age, n in cell_metadata.groupby("donor_age")["donor_label"].nunique().items()
    if n >= 2
]

print(len(valid_ages))
print(valid_ages)
```

```python
import os
from itertools import combinations

os.makedirs("/content/DE_subtables", exist_ok=True)

for age1, age2 in combinations(valid_ages, 2):
    idx1 = donor_info["donor_age"] == age1
    idx2 = donor_info["donor_age"] == age2

    _, p_values = ttest_ind(
        log_cpm[idx1],
        log_cpm[idx2],
        axis=0,
        equal_var=False
    )

    p_values_clean = np.nan_to_num(p_values, nan=1.0)
    fdr = multipletests(p_values_clean, method="fdr_bh")[1]

    de_table = pd.DataFrame({
        "age_1": age1,
        "age_2": age2,
        "gene_identifier": gene_metadata["gene_identifier"].values,
        "gene_symbol": gene_metadata["gene_symbol"].values,
        "p_value": p_values,
        "FDR": fdr
    })

    filename = f"DE_{age1.replace(' ', '_')}_vs_{age2.replace(' ', '_')}.csv"
    de_table.to_csv(f"/content/DE_subtables/{filename}", index=False)

print("Completed.")
print("Number of subtables:", len(os.listdir("/content/DE_subtables")))
```

```python
import glob
import pandas as pd
import os

summary = []

for file in glob.glob("/content/DE_subtables/*.csv"):
    df = pd.read_csv(file)

    summary.append({
        "comparison": os.path.basename(file).replace(".csv", ""),
        "age_1": df["age_1"].iloc[0],
        "age_2": df["age_2"].iloc[0],
        "genes_tested": len(df),
        "significant_genes_FDR_0.05": (df["FDR"] < 0.05).sum()
    })

summary_df = pd.DataFrame(summary)

summary_df.to_csv(
    "/content/DE_comparison_summary.csv",
    index=False
)

print("Completed.")
print("Comparisons:", len(summary_df))
print("Saved: /content/DE_comparison_summary.csv")
```
