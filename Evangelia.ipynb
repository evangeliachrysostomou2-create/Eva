```python
import numpy as np
from scipy import sparse

# Donor information in exactly the same cell order as adata
donor_info = cell_donor.loc[adata.obs_names]

# Unique donors and numeric code for each cell
donors = donor_info["donor_label"].unique()
donor_to_idx = {d: i for i, d in enumerate(donors)}
donor_idx = donor_info["donor_label"].map(donor_to_idx).to_numpy()

# Matrix that will contain summed counts:
# rows = donors, columns = genes
donor_counts = np.zeros((len(donors), adata.n_vars), dtype=np.float64)

# Read the h5ad in chunks
for X_chunk, start, end in adata.chunked_X(10000):
    codes = donor_idx[start:end]

    if sparse.issparse(X_chunk):
        group_matrix = sparse.csr_matrix(
            (
                np.ones(len(codes)),
                (codes, np.arange(len(codes)))
            ),
            shape=(len(donors), len(codes))
        )
        donor_counts += group_matrix @ X_chunk
    else:
        for d in np.unique(codes):
            donor_counts[d] += np.asarray(X_chunk)[codes == d].sum(axis=0)

    if end % 100000 == 0:
        print(f"Processed {end:,} / {adata.n_obs:,} cells")

print("Done:", donor_counts.shape)
```

```python
from scipy.stats import ttest_ind
from statsmodels.stats.multitest import multipletests
import pandas as pd
import numpy as np

# Keep only age groups with >=2 donors
valid_ages = [
    age for age in donor_metadata["donor_age"].unique()
    if (donor_metadata["donor_age"] == age).sum() >= 2
]

results = []

for i in range(len(valid_ages)):
    for j in range(i + 1, len(valid_ages)):

        age1 = valid_ages[i]
        age2 = valid_ages[j]

        idx1 = np.where(donor_metadata["donor_age"].values == age1)[0]
        idx2 = np.where(donor_metadata["donor_age"].values == age2)[0]

        stat, pval = ttest_ind(
            log_cpm[idx1],
            log_cpm[idx2],
            axis=0,
            equal_var=False
        )

        padj = multipletests(pval, method="fdr_bh")[1]

        effect = (
            log_cpm[idx1].mean(axis=0)
            - log_cpm[idx2].mean(axis=0)
        )

        tmp = pd.DataFrame({
            "age_1": age1,
            "age_2": age2,
            "gene_identifier": gene["gene_identifier"].values,
            "gene_symbol": gene["gene_symbol"].values,
            "mean_log1pCPM_difference": effect,
            "p_value": pval,
            "FDR": padj
        })

        results.append(tmp)

        print(f"{age1} vs {age2} done")

de_catalog = pd.concat(results, ignore_index=True)

print("DONE")
print("Comparisons:", de_catalog[["age_1", "age_2"]].drop_duplicates().shape[0])
print("Rows:", len(de_catalog))
```
